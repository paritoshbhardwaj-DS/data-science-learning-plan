```python
import numpy

print(numpy.zeros((1,2)))
print((numpy.zeros((1,2), dtype = int)))
```

```python
print(numpy.ones((1,2)))
print(numpy.ones((1,2), dtype = int))
```

```python
shape = tuple(map(int, input().split()))

print(numpy.zeros(shape, dtype=int))
print(numpy.ones(shape, dtype=int))
```

Performance: Tuples are slightly faster and use less memory than lists in Python because they are immutable (cannot be changed). However, for a small script like a HackerRank challenge, the difference is measured in nanoseconds—completely unnoticeable.
