```python
import numpy

a = numpy.array([1,2,3,4], float)
b = numpy.array([5,6,7,8], float)
print(a+b)
print(numpy.add(a,b))
print(a-b)
print(numpy.subtract(a,b))
print(a*b)
print(numpy.multiply(a,b))
print(a/b)
print(numpy.divide(a,b))
print(a%b)
print(numpy.mod(a,b))
print(a**b)
print(numpy.power(a,b))
```
